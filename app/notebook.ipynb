```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Create a sample dataset with missing values
data = {
    'Age': [25, 30, 35, 40, 45],
    'ownedHouse': [True, False, True, False, True],
    'creditScore': [700, 650, 720, 680, 710],
    'loanAmount': [200000, 150000, 250000, 180000, 220000],
    'Marital Status': ['Single', 'Married', 'Single', None, 'Single'],
    'Education Level': ['Bachelor', 'Master', 'PhD', 'Bachelor', 'Master'],
    'Empoyment Status': ['Full-time', 'Part-time', 'Unemployed', 'Full-time', 'Part-time'],
    'Loan Approval': ['Yes', 'No', 'Yes', 'No', 'Yes']
}
```

```python
# Crate a DataFrame
df = pd.DataFrame(data)
# Display the DataFrame
df.head(10)
```

```python
# info about the dataset
df.info()
```

```python
# get the data types of the columns
df.dtypes
```

```python
# statistical summary of the dataset
df.describe()
```

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing values in each column:")
print(missing_values)
```

```python
# Check the duplicate values
duplicate_values = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_values)   
```

```python
# Check the unique values in the 'Marital Status' column
unique_marital_status = df['Marital Status'].unique()
print("Unique values in 'Marital Status' column:", unique_marital_status)
```

```python
# Correlation matrix
correlation_matrix = df[['Age', 'creditScore', 'loanAmount']].corr()
correlation_matrix
```

```python
# Heatmap of the correlation matrix
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix Heatmap')
plt.show() 
```

```python
# Visualize the distribution of 'Age' using a histogram
plt.figure(figsize=(8, 6))
sns.histplot(df['Age'], bins=10, kde=True)
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# Visualize the distribution of 'loanAmount' using a histogram
plt.figure(figsize=(8, 6))
sns.histplot(df['loanAmount'], bins=10, kde=True)
plt.title('Distribution of Loan Amount')
plt.xlabel('Loan Amount')
plt.ylabel('Frequency')
plt.show()
```

```python
# Boxplot to visualize the relationship between 'Education Level' and 'loanAmount'
plt.figure(figsize=(10, 6))
sns.boxplot(x='Education Level', y='loanAmount', data=df)
plt.title('Loan Amount by Education Level')
plt.xlabel('Education Level')
plt.ylabel('Loan Amount')
plt.show()
```

```python
# Pairplot to visualize relationships between numerical variables
sns.pairplot(df[['Age', 'creditScore', 'loanAmount']])
plt.suptitle('Pairplot of Age, Credit Score, and Loan Amount', y=1.02)
plt.show()
```
