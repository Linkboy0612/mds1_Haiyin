```python
# import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# load the dataset
df = pd.read_csv('../Data/Simple_Linear_Regression.csv')
# display the dataframe
df.head()
```

```python
# print the missing values in the dataset
print(df.isnull().sum())
```

```python
# display descriptive statistics of the dataset
df.describe()
```

```python
# scatter plot of the dataset
plt.scatter(df['Year'], df['Per capita income (E)'])
plt.xlabel('Year')
plt.ylabel('Per capita income (E)')
plt.title('Scatterplot of x and y')
plt.show()
```

```python
# create the feature matrix (X) and target vector (y)
X = df[['Year']]
y = df['Per capita income (E)']
```

```python
# training the linear regression model
model = LinearRegression()
model.fit(X, y)
# make predictions using the trained model
y_pred = model.predict(X)
```

```python
# model parameters
print("Slope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)
# y = mx + b
```

```python
# evaluate the model
rmse = np.sqrt(mean_squared_error(y, y_pred))
r2 = r2_score(y, y_pred)
print("Root Mean Squared Error (RMSE):", rmse)
print("R-squared (R2):", r2)
```

```python
# predict the per capita income for the year 2027
year_2027 = pd.DataFrame([[2027]], columns=['Year'])
predicted_income_2027 = model.predict(year_2027)
print(f"Predicted per capita income for the year 2027: {predicted_income_2027[0]:.2f} ± {rmse:.2f}")
```

```python
# plot the regression line
plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', label='Actual data')
plt.plot(X, y_pred, color='red', label='Regression line')
plt.xlabel('Year')
plt.ylabel('Per capita income (E)')
plt.title('Linear Regression: Year vs Per capita income (E)')
plt.legend()
plt.grid(True)
plt.show()
```
